# Latest Forecast vs Dengue Predictor (short form)

## Latest Forecast

- The prediction is made **in the notebook**, then imported into the dashboard.
- The notebook takes the last week in the dataset (July 2024) and builds all 24 features from **real data**.
- The already trained models predict the following week (no retraining).
- The results are saved to `forecast_next_week.csv`. The app **only reads and displays** this file.
- The app adds the risk level (district's own p50 / p80) and the likely range (from validation errors).
- The forecast does not change until the notebook is re-run with newer data.

## Dengue Predictor

- It uses the **same trained models** (Random Forest, Stacking, Linear Regression), run **live in the app**.
- The user enters the district, the week, cases for 5 weeks, and a rainfall setting.
- The 24 features are built as follows:
  - 4 are typed by the user.
  - 1 is calculated from the user's cases.
  - 3 are calculated from the district and week.
  - 16 are filled with district-week averages.
- Averages are calculated from all years (2007 to July 2024).

## Why averages are used

- Most of the 24 features are **not dataset columns**. They are calculated (lags, rolling totals).
- For the Latest Forecast we have the full dataset, so the notebook can calculate them easily.
- A normal user does not have the weekly history, so calculating all 24 by hand is not practical.
- For this assignment, a simplified what-if tool asks for a few inputs and fills the rest with averages.
- It is not a validated forecast. A backtest (training-year averages only, no test data) measures the accuracy cost.

## Why several weeks before, but only one week after?

- **Why we use several past weeks (lags):**
  - Dengue spreads with delays. Mosquito eggs and larvae develop, the virus incubates in the mosquito and in people, and cases are reported late.
  - Because of these delays, cases and rainfall from earlier weeks carry information about next week.
  - So the models use case lags (1, 2, 4, 8, 12 weeks) and rainfall lags (2 to 10 weeks).
- **Why only one week ahead:**
  - The models were trained on one task: use the weeks up to now to predict **next week**.
  - They rely mostly on **recent cases** (this week, last week, 4-week average).
  - For a week several weeks after the data end, those recent weeks have not happened, so their values are unknown.
  - Past weeks help to predict the next week, but the next week's cases are unknown for the week after that.
- **To go further ahead (not done in this study):**
  - Recursive: use each prediction as input for the next step (errors add up).
  - Direct: train separate models for 2, 3, 4 weeks ahead, then tune and test them.
- **To keep forecasting:** add the new real weekly data, rebuild the features, and re-run the forecast step. The saved models can be reused.
- **Caution:** the data show that rainfall is linked most strongly about 8 to 9 weeks later. This is longer than the mosquito life cycle alone, so part of it may be seasonality. These are associations, not proof of cause.
